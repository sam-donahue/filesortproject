```python
# ASSUMPTIONS:
# Current date is always 4/1/2026
# When creating a file, dslo is zero.

# ~~FUNCTIONS~~

#validates data based on phrase and max value

def datavalidator(phrase, maxnum):
    temp = list(range(1, maxnum + 1))
    while True:
        num = input(phrase)
        try:
            val = int(num)
            if val in temp:
                return val
            else:
                if maxnum == 2:
                    print("Integer out of bounds. Please enter 1 or 2.")
                else:
                    print(f"Integer out of bounds. Please enter a number from 1 to {maxnum}.")
        except ValueError:
            print("Value is not an integer. Please select a valid option.")

# Lets the user decide to go to file sorting or add files first.
            
def mainmenu():
    print("1- Sort Files\n2- Add File(s)")
    selection = datavalidator("Enter a selection: ", 2)
    match selection:
        case 1:
            filesorter()
            sortby()
        case 2:
            fileappender()

#converts lines on the text file to a list of dictionaries

def filesorter():
    samplelist = []
    with open('samplefiles.txt', 'r') as file:
        for line in file:
            line = line.strip()
            presort = line.split('.')
            entry = {
                "date": presort[0],
                "dotw": presort[1],
                "name": presort[2],
                "title": presort[3],
                "size": presort[4],
                "dslo": presort[5]
            }
            samplelist.append(entry)
    print(samplelist)
    return samplelist

# tells program what to sort by

def sortby():
    sortans = input("What value would you like to sort by?\n\n1 - Date\n2 - Name\n3 - Title\n")
    return sortans

# appends lines to the file based on user imput.

def fileappender():
    valans = datavalidator("How many files would you like to add? ", 100)
    print("Adding ", valans, " files.")
    for i in range(valans):
        with open('samplefiles.txt','a') as file:
            month = datavalidator("Enter the creation month (1-12): ", 12)
            match month:
                case 1 | 3 | 5 | 7 | 8 | 10 | 12:
                    day = datavalidator("Enter the creation day (1-31): ", 31)
                case 4 | 6 | 9 | 11:
                    day = datavalidator("Enter the creation day (1-30): ", 30)
                case 2:
                    day = datavalidator("Enter the creation day (1-29): ", 29)
            year = datavalidator("Enter the creation year (2026 or earlier): ", 2026)
            author = input("Enter the author's name: ")
            tempsubject = datavalidator("Enter the subject.\n1- Algebra\n2- Programming\n3- English\n", 3)
            match tempsubject:
                case 1:
                    subject = "algebra"
                case 2:
                    subject = "programming"
                case 3:
                    subject = "english"
            tempwork = datavalidator("Enter the type of work.\n1- Notes\n2- Homework\n3- Study Guide\n", 3)
            match tempwork:
                case 1:
                    work = "notes"
                case 2:
                    work = "homework"
                case 3:
                    work = "studyguide"
            size = datavalidator("Enter file size (1-999): ", 999)
            dslo = "0"
            line = f"{month:02}/{day:02}/{year}.{author}.{subject}_{work}.{size}.{dslo}\n"
            file.write(line)
    returnans = datavalidator("Would you like to return to the main menu?\n1- Yes\n2- No\n", 2)
    if returnans == 1:
        mainmenu()

# ~~MAIN~~

mainmenu()
```
